# Titanic MVC en Jupyter

Arquitectura:

```text
            JUPYTER
               |
               v
         Controller
         /        \
        v          v
      Model <---- View
        |
        v
   DataFrame
        |
        v
titanic4_2026.csv
```

- **Model:** contiene y utiliza el objeto `DataFrame`.
- **View:** contiene únicamente widgets y salida visual.
- **Controller:** recibe Model y View, escucha eventos y ejecuta la lógica.
- **Jupyter:** actúa únicamente como programa principal.


````python
# Si hiciera falta:
# %pip install ipywidgets

from model import TitanicModel
from view import TitanicView
from controller import TitanicController
````

````python
# MODEL
model = TitanicModel("drug200.csv")

# VIEW
view = TitanicView()

# CONTROLLER
controller = TitanicController(model, view)
````
