```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets

from IPython.display import display, clear_output
from DataFrame_desarrollado import DataFrame

RUTA = "drug200.csv"
base = pd.read_csv(RUTA)

print("Dataset cargado correctamente")
print("Dimensiones:", base.shape)
display(base.head())
```

```python

# ============================================================
# PRUEBAS DE TODOS LOS MÉTODOS DE LA CLASE
# ============================================================

METODOS = [
    "__init__ / __str__",
    "propiedad dataframe",
    "setter dataframe",
    "cargar_csv",
    "mostrar",
    "primeras_filas",
    "ultimas_filas",
    "recorrer_filas",
    "recorrer_columnas",
    "obtener_fila",
    "obtener_columna",
    "dimensiones",
    "nombres_columnas",
    "tipos_datos",
    "contar_nulos",
    "eliminar_nulos",
    "reemplazar_nulos",
    "ordenar",
    "filtrar",
    "agregar_columna",
    "eliminar_columna",
    "estadisticas",
    "correlacion",
    "guardar_csv",
    "visualizar_datos",
    "frecuencias",
    "detectar_duplicados",
    "eliminar_duplicados",
    "convertir_tipos",
    "distribucion_variables",
    "histogramas",
    "boxplots",
    "scatterplots",
    "mapa_calor",
    "detectar_outliers",
    "imputar_nulos",
    "normalizar",
    "estandarizar",
    "seleccionar_variables",
    "exportar_resultados",
]

def nueva_instancia(muestra=None):
    datos = base.copy() if muestra is None else base.head(muestra).copy()
    return DataFrame(datos)

def titulo(nombre):
    print("=" * 72)
    print(f"PRUEBA: {nombre}")
    print("=" * 72)

def ejecutar_prueba(nombre):
    titulo(nombre)

    if nombre == "__init__ / __str__":
        df = DataFrame(base)
        print(df)

    elif nombre == "propiedad dataframe":
        df = nueva_instancia()
        display(df.dataframe.head())

    elif nombre == "setter dataframe":
        df = nueva_instancia()
        copia = base.head(10).copy()
        df.dataframe = copia
        print("Nuevas dimensiones:", df.dimensiones())
        display(df.dataframe)

    elif nombre == "cargar_csv":
        df = DataFrame()
        resultado = df.cargar_csv(RUTA)
        print("Dimensiones cargadas:", resultado.shape)
        display(resultado.head())

    elif nombre == "mostrar":
        df = nueva_instancia()
        display(df.mostrar().head(20))

    elif nombre == "primeras_filas":
        df = nueva_instancia()
        display(df.primeras_filas(10))

    elif nombre == "ultimas_filas":
        df = nueva_instancia()
        display(df.ultimas_filas(10))

    elif nombre == "recorrer_filas":
        df = nueva_instancia(muestra=3)
        df.recorrer_filas()

    elif nombre == "recorrer_columnas":
        df = nueva_instancia(muestra=5)
        df.recorrer_columnas()

    elif nombre == "obtener_fila":
        df = nueva_instancia()
        print(df.obtener_fila(0))

    elif nombre == "obtener_columna":
        df = nueva_instancia()
        display(df.obtener_columna("age").head(20).to_frame())

    elif nombre == "dimensiones":
        df = nueva_instancia()
        print(df.dimensiones())

    elif nombre == "nombres_columnas":
        df = nueva_instancia()
        print(df.nombres_columnas())

    elif nombre == "tipos_datos":
        df = nueva_instancia()
        print(df.tipos_datos())

    elif nombre == "contar_nulos":
        df = nueva_instancia()
        display(df.contar_nulos().to_frame("nulos"))

    elif nombre == "eliminar_nulos":
        df = nueva_instancia()
        antes = df.dimensiones()
        resultado = df.eliminar_nulos()
        print("Antes:", antes, "| Después:", resultado.shape)
        display(resultado.head())

    elif nombre == "reemplazar_nulos":
        df = nueva_instancia()
        resultado = df.reemplazar_nulos(0)
        print("Nulos después:", resultado.isna().sum().sum())
        display(resultado.head())

    elif nombre == "ordenar":
        df = nueva_instancia()
        resultado = df.ordenar("age")
        display(resultado.head(15))

    elif nombre == "filtrar":
        df = nueva_instancia()
        resultado = df.filtrar("survived", "==", 1)
        print("Sobrevivientes:", len(resultado))
        display(resultado.head(15))

    elif nombre == "agregar_columna":
        df = nueva_instancia()
        resultado = df.agregar_columna("prueba", 1)
        display(resultado[["survived", "prueba"]].head())

    elif nombre == "eliminar_columna":
        df = nueva_instancia()
        resultado = df.eliminar_columna("fare")
        print("Columnas:", list(resultado.columns))
        display(resultado.head())

    elif nombre == "estadisticas":
        df = nueva_instancia()
        display(df.estadisticas().T)

    elif nombre == "correlacion":
        df = nueva_instancia()
        display(df.correlacion())

    elif nombre == "guardar_csv":
        df = nueva_instancia()
        ruta = df.guardar_csv("titanic_guardado_prueba.csv")
        print("Archivo creado:", ruta)

    elif nombre == "visualizar_datos":
        df = nueva_instancia()
        resultado = df.visualizar_datos()
        display(resultado)

    elif nombre == "frecuencias":
        df = nueva_instancia()
        frecs = df.frecuencias()
        for columna, tabla in frecs.items():
            print(f"\n--- {columna} ---")
            display(tabla.head(10).to_frame("frecuencia"))

    elif nombre == "detectar_duplicados":
        df = nueva_instancia()
        resultado = df.detectar_duplicados()
        print("Filas duplicadas detectadas:", len(resultado))
        display(resultado.head(20))

    elif nombre == "eliminar_duplicados":
        df = nueva_instancia()
        antes = df.dimensiones()
        resultado = df.eliminar_duplicados()
        print("Antes:", antes, "| Después:", resultado.shape)
        display(resultado.head())

    elif nombre == "convertir_tipos":
        df = nueva_instancia()
        resultado = df.convertir_tipos({"pclass": "float64"})
        print(resultado.dtypes)
        display(resultado.head())

    elif nombre == "distribucion_variables":
        df = nueva_instancia()
        resultado = df.distribucion_variables()
        print("NUMÉRICAS")
        display(resultado["numericas"])
        print("CATEGÓRICAS")
        display(resultado["categoricas"])

    elif nombre == "histogramas":
        df = nueva_instancia()
        df.histogramas()

    elif nombre == "boxplots":
        df = nueva_instancia()
        df.boxplots()

    elif nombre == "scatterplots":
        df = nueva_instancia()
        df.scatterplots()

    elif nombre == "mapa_calor":
        df = nueva_instancia()
        matriz = df.mapa_calor()
        display(matriz)

    elif nombre == "detectar_outliers":
        df = nueva_instancia()
        resultado = df.detectar_outliers()
        resumen = {col: len(filas) for col, filas in resultado.items()}
        display(pd.Series(resumen, name="cantidad_outliers").to_frame())

    elif nombre == "imputar_nulos":
        df = nueva_instancia()
        antes = int(df.dataframe.isna().sum().sum())
        resultado = df.imputar_nulos()
        despues = int(resultado.isna().sum().sum())
        print("Nulos antes:", antes, "| Nulos después:", despues)
        display(resultado.head())

    elif nombre == "normalizar":
        df = nueva_instancia()
        resultado = df.normalizar()
        display(resultado.head())
        display(resultado.select_dtypes(include=np.number).agg(["min", "max"]))

    elif nombre == "estandarizar":
        df = nueva_instancia()
        resultado = df.estandarizar()
        display(resultado.head())
        display(resultado.select_dtypes(include=np.number).agg(["mean", "std"]).round(4))

    elif nombre == "seleccionar_variables":
        df = nueva_instancia()
        resultado = df.seleccionar_variables(["pclass", "survived", "sex", "age"])
        display(resultado.head(20))

    elif nombre == "exportar_resultados":
        df = nueva_instancia()
        ruta = df.exportar_resultados("resultados_eda_titanic.csv")
        print("Archivo creado:", ruta)

    print("\n✓ Prueba finalizada")

selector = widgets.Dropdown(
    options=METODOS,
    value=METODOS[0],
    description="Método:",
    layout=widgets.Layout(width="420px")
)

btn_probar = widgets.Button(description="Probar método", button_style="primary")
btn_todos = widgets.Button(description="Probar todos")
salida = widgets.Output()

def click_probar(_):
    with salida:
        clear_output(wait=True)
        try:
            ejecutar_prueba(selector.value)
        except Exception as e:
            print(f"ERROR: {type(e).__name__}: {e}")

def click_todos(_):
    with salida:
        clear_output(wait=True)
        for nombre in METODOS:
            try:
                ejecutar_prueba(nombre)
            except Exception as e:
                print(f"ERROR en {nombre}: {type(e).__name__}: {e}")
            print("\n")

btn_probar.on_click(click_probar)
btn_todos.on_click(click_todos)

display(widgets.HTML("<h2>Prueba de métodos de la clase DataFrame</h2>"))
display(widgets.HBox([selector, btn_probar, btn_todos]))
display(salida)
```
