# MVC de análisis de datos en Jupyter

Estructura del proyecto:

```text
IA_Laboratorios/
├── data/                    datasets CSV
├── src/
│   ├── dataframe_desarrollado.py  EDA y adaptador de análisis
│   ├── model.py, view.py, controller.py
│   ├── unsupervised/        PCA, HAC, K-Means, t-SNE y UMAP
│   └── supervised/          espacio para futuros modelos supervisados
└── notebooks/               este notebook
```

Arquitectura:

```text
            JUPYTER
               |
               v
        DataController
         /              \
        v                v
    DataModel <------ DataView
        |
        v
   DataFrame (EDA compartido)
        |
        +-- unsupervised/   reducción dimensional y clustering
        +-- supervised/     clasificación y regresión (futuro)
```

- **DataModel:** carga cualquier CSV, restaura y delega las operaciones a `DataFrame`.
- **DataFrame:** concentra EDA común y conserva una API única para el MVC.
- **unsupervised:** contiene la implementación de PCA, HAC, K-Means, t-SNE y UMAP.
- **supervised:** paquete separado para agregar modelos supervisados sin mezclar responsabilidades.
